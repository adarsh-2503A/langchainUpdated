### pydantic structured output

```python
import os
from dotenv import load_dotenv

load_dotenv()
api_key=os.getenv('GROQ_API_KEY')
if not api_key:
    raise ValueError('Api key not found in env file')
```

```python
from pydantic import BaseModel
from datetime import datetime

class Ticket(BaseModel):
    name:str
    issue:str
    date: datetime
```

```python
from groq import Groq
import json

schema= Ticket.model_json_schema()
response_format={"type":"json_object"}
system_prompt={"role":"system","content":f'Give the json response strictly in this format {schema}'}
text="Hi my name is Shivani I am from Hyderabad, I am a great singer, i have an issue with my laptop." \
" The issue is present from 12th may of this year"
prompt={"role":"user","content":f"Extract information from this {text}"}

model='openai/gpt-oss-120b'
client=Groq(api_key=api_key)
response=client.chat.completions.create(messages=[system_prompt,prompt],model=model,response_format=response_format)
response=response.choices[0].message.content


data=json.loads(response)
ticket=Ticket(**data)
print(ticket)
print(ticket.name)
print(ticket.date)
```
