```python
import os
from dotenv import load_dotenv

load_dotenv()
api_key=os.getenv('GROQ_API_KEY')
if not api_key:
    raise ValueError('Api key not found in env file')
```

```python
from groq import Groq

model='openai/gpt-oss-120b'
client=Groq(api_key=api_key)
response=client.chat.completions.create(messages=[{"role":"user","content":"Explain the big bang theory in 2 lines"}],model=model)
print(response.choices[0].message.content)
```

```python
system_message={"role":"system","content":"you are a brand manager who suggests name for my business, suggest only one name"}
response=client.chat.completions.create(messages=[system_message,{"role":"user","content":"suggest a name for my food company"}],model=model,temperature=1)
print(response.choices[0].message.content)
```

```python
prompts=["hi","explain about the company which pioneered Ai usage in daily life"]

for prompt in prompts:
    response=client.chat.completions.create(messages=[{"role":"user","content":prompt}],
                                            model=model,max_tokens=500)
    print(response.choices[0].message.content)
    print("Stop reason: ", response.choices[0].finish_reason)
    print("*"*50)
```
